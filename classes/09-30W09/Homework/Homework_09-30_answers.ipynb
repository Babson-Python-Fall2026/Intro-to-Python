````python
# (a) Trace table (fill in BEFORE running):
# amount | action (added / skipped / stopped) | total after | count after
#   120  |            added                   |     120     |1
#    -1  |            skipped                 |     120     |1
#    45  |            added                   |     165     |2
#     0  |            skipped                 |     165     |2
#   300  |            added                   |     465     |3
#    -1  |            skipped                 |     465     |3
#    80  |            added                   |     545     |4
#   999  |            stopped                 |     545     |4

#
# (b) Printed       :545 4
#     Never examined: 60

orders = [120, -1, 45, 0, 300, -1, 80, 999, 60]
total = 0
count = 0

for amount in orders:
    if amount == 999:
        break
    if amount <= 0:
        continue
    total += amount
    count += 1

print(total, count)
````

### Part C: PIN lockout

Write a program that asks for a PIN using `input()`. The correct PIN is `"4821"`. The user gets at most 3 attempts:

- Correct PIN: print `Access granted` and stop asking.
- Wrong PIN: print how many attempts are left.
- After 3 wrong attempts: print `Card locked`.

Test it twice: once succeeding on the 2nd try, once failing 3 times. The notebook only keeps the output of the last run, so paste the output of the first run in the comment.

````python
# your code here

correct_pin = "4821"
count = 0
while True:
    pin = input("Please input your pin: ")
    count = count + 1
    if pin == correct_pin:
        print('Access granted')
        break
    else:
        print(3 - count,'tries left')
    if count == 3:
        print("Card locked")
        break

# Output of test 1 (success on 2nd try):
# 2 tries left
# Access granted
````

### Part C: discount code validator

Define a custom exception `InvalidDiscountError`. Then write a loop that keeps asking for a discount percentage until it gets a valid one:

- Not an integer: print an error and ask again.
- An integer outside 0–50: `raise InvalidDiscountError` with a helpful message, print that message, and ask again.
- Valid: in the `else` block, print the sale price of an \$80 item after the discount, and stop.

Run it once, entering `ten`, `75`, `-5`, `20` in that order.

````python
# your code here


class InvalidDiscountError(Exception):
    pass

while True:
    try:
        discount = int(input('Please enter a discount percentage: '))
        if not (0 <= discount <= 50):
            raise InvalidDiscountError('discount out of range, try again\n\n')
    except ValueError:
        print('Bad input, try again\n\n')  
    except InvalidDiscountError as e:
        print(e)    
    else:
        print(f'${80 * (100 - discount) / 100:.2f}') 
        break           
````
