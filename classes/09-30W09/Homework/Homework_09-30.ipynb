## Problem 5: Trace `continue` and `break` (10 pts)

An order log uses `-1` and `0` for voided entries and `999` as an end-of-day marker.

**(a)** *Without running*, fill in the trace table in the cell below, one row per iteration.
**(b)** What is printed? Which value in the list is never examined at all?

````python
# (a) Trace table (fill in BEFORE running):
# amount | action (added / skipped / stopped) | total after | count after
#   120  |                                    |             |
#        |                                    |             |
#
# (b) Printed:
#     Never examined:

orders = [120, -1, 45, 0, 300, -1, 80, 999, 60]
total = 0
count = 0

for amount in orders:
    if amount == 999:
        break
    if amount <= 0:
        continue
    total += amount
    count += 1

print(total, count)
````
