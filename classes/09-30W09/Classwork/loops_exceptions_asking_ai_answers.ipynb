## 4. Exceptions

Some operations can fail even when the program's logic is otherwise reasonable. A `try` statement lets us attempt an operation and handle specific exceptions.

The complete structure is:

```python
try:
    # code that might fail
except SomeError:
    # runs if that error occurs
else:
    # runs only if no exception occurred
finally:
    # runs whether or not an exception occurred
```

### Common Python exceptions

| Exception | What it means | Simple example |
|---|---|---|
| `ValueError` | The value is not acceptable for the operation. | `int("abc")` |
| `TypeError` | An operation is used with the wrong type of object. | `"5" + 3` |
| `IndexError` | A sequence index does not exist. | `[10, 20][5]` |
| `KeyError` | A dictionary key does not exist. | `{"a": 1}["b"]` |
| `NameError` | A variable or name has not been defined. | `print(score)` |
| `ZeroDivisionError` | Division or modulo by zero was attempted. | `10 / 0` |
| `FileNotFoundError` | Python cannot find the requested file. | `open("missing.txt")` |


````python
'''
Predict what runs when the user enters:

1. 25
2. hello

Pay attention to which blocks execute:
try
except
else
finally
'''

try:
    age = int(input("Enter your age: "))
except ValueError:
    print("except: please enter a whole number")
else:
    print("else: conversion succeeded")
    print("age:", age)
finally:
    print("finally: this always runs")
````

### Raising an exception yourself

Python raises many exceptions automatically, but your own program can also use `raise` when a value violates a rule.

For example:

```python
raise ValueError("message")
```

The minimal pattern for defining a custom exception is:

```python
class MyError(Exception):
    pass
```

Then you can raise it with:

```python
raise MyError("some message")
```


````python
'''
This program requires an integer from 1 through 10.

Predict what happens for:
5
0
11
hello

Notice that ValueError can come from:
1. int() if conversion fails
2. our own raise statement if the number is outside the allowed range
'''

class BoundsCheck(Exception):
    pass

try:
    number = int(input("Enter an integer from 1 to 10: "))

    if number < 1 or number > 10:
        raise BoundsCheck

except ValueError as error:
    print("Invalid input:", error)
except BoundsCheck:
    print('Value out of range')
else:
    print("Accepted:", number)
finally:
    print("Input check complete.")
````

## 6. Asking AI: Incomplete vs. Complete Specifications

When you ask AI to write a program, **every decision your request doesn't make, the AI makes for you.** Two requests that sound almost the same can leave very different decisions open, and produce different programs.

There are layers of control:

| Layer | Who decides |
|---|---|
| **Problem** — what needs to be solved | You |
| **Specification** — the steps and the important design choices | You |
| **Implementation** — anything your specification left open | The AI |
| **Execution** — exactly what the code does | Python |

The goal is **not** to specify every line of code. That is just writing Python in English. The goal is to make the decisions that *matter* yourself, and to know which ones you are handing to the AI.

Decisions that usually matter:

- **Data representation** — list, dictionary, tuple? What are the keys?
- **Input** — where does the data come from, and in what form?
- **Stopping** — how does the program know it is finished?
- **Bad input** — text where a number belongs, negative values, unknown items
- **Edge cases** — nothing entered, zero, ties, running out
- **Rules** — limits, rounding, what counts as what
- **Output** — what is reported, and in what form

### Two habits for working with AI

**1. A clarifying question is a missing decision.**
When the AI asks you something ("What should happen if the user enters text?"), you left that decision open. Don't just answer in the chat. Add the answer to your specification.

**2. Ask the decisions question.**
After the AI writes code, ask:

> List every decision you made that my specification didn't state.

For each item on the list, decide: does this matter to me? If it does, put it in your specification. If it doesn't, you have knowingly handed it to the AI.

Your specification is **complete** when the AI has nothing left to ask, and every decision left on its list is one you are comfortable letting it make.
