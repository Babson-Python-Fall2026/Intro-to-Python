# Class 2 · Working with AI — In-Class Exercises
## Poor Prompts vs. Good Prompts: Ten Experiments

Today you run **experiments on the AI**, not exercises about it. Every experiment has the same rhythm:

1. **📤 Send the poor prompt** — yes, on purpose. Copy it exactly as written.
2. **🔮 Predict** — before sending the good version, write down what you expect to change.
3. **📥 Send the good prompt** — then compare what actually came back.
4. **🗒️ Log it** — capture the difference in the answer box.
5. **👥 Compare** — most exercises have a pair or whole-room moment. The differences *between your screens* are half the lesson.

**Setup check (do this now):** open Claude in a browser tab next to this notebook. For Exercises 1–9, use a **regular chat** (NOT your OIM3600 project — we want plain, unscaffolded Claude today so you can see the raw behavior). Exercise 10 moves into your project.

You don't need to know any Python for this class. One exercise hands you three lines of code to *run* — reading them requires nothing but arithmetic.

⏱️ Pace: about 7–8 minutes per exercise. If one runs long, your instructor will call time — the log boxes save your place.

## Exercise 3 — Name Your Intent ⏱️ 8 min

The workbook's Part 3 lesson: the same topic can be asked for **as a different kind of answer** — and if you don't choose, the AI chooses for you.

### 📥 Send all three of these — same topic, three intents — as three separate messages

## Exercise 5 — Never Accept the First Answer ⏱️ 8 min

The detective drill. The poor version of this exercise is what most people do every day: **ask once, copy, leave.**

### 📥 Step 1 — Ask something real
Ask Claude any genuine how-should-I question — study habits, gym routine, budgeting, time management. One sentence is fine.

### 📥 Step 2 — Now push back, three rounds, using the Detective Toolkit
Send these as three separate follow-ups:
```
1) How confident are you in that advice? What could make it wrong
   for someone like me?

2) Show me a situation where that advice would backfire.

3) Give me one completely different approach and the tradeoff
   between the two.
```


> 🗒️ **Log it:**
>
> What round 3's answer contained that round 0's didn't: …
>
> Honestly: would I have caught any of that if I'd stopped after the first answer? …
>
> 💡 **Lock it in:** *"The first answer is a ______, not a ______."*

## Exercise 6 — The Hallucination Hunt ⏱️ 8 min

AI sounds equally confident when it's right and when it's wrong. Time to catch it in the act — or fail to, which is also data.

### 📤 Send the bait prompt
```
Give me three exact, word-for-word quotes from famous business
leaders about failure. Include who said it and where (book, speech,
or interview).
```

### 🔍 Now verify — this is the actual exercise
Pick quote #1. Search the web for the exact phrase in quotation marks. Three possible outcomes: **verified** (real quote, right source), **mangled** (real idea, wrong words or wrong source), or **fabricated** (no trace of it anywhere).

### 👥 Whole-room moment
Instructor tallies the room: how many verified / mangled / fabricated? (Most semesters, the room catches at least a few.)


> 🗒️ **Log it:**
>
> My quote #1 verdict (verified / mangled / fabricated) and the evidence: …
>
> The dangerous part wasn't the error — it was the ______ of the error. (Hint: how did the answer *sound*?)
>
> One thing I will now always do before putting an AI "fact" in a paper: …

## Exercise 8 — The Three Approaches, Live ⏱️ 10 min · the centerpiece

The workbook's core idea, now run as a controlled experiment. **One problem, three conversations.** Use a fresh chat for each so they can't contaminate each other.

**The problem:** *Three roommates share a $2,400/month apartment. The bedrooms are different sizes: 200, 150, and 130 square feet, plus shared common space. How should rent be split fairly?*

### 📤 Conversation A — Approach 1: Outsource the Answer
```
Three roommates pay $2400 rent, bedrooms are 200, 150 and 130 sq ft.
What should each pay? Just give me the numbers.
```

### 📤 Conversation B — Approach 2: Outsource the Design
```
Design a fair method for splitting $2400 rent among three roommates
with bedrooms of 200, 150, and 130 sq ft plus shared common areas.
Choose the method you think is best and show the result.
```

### 📥 Conversation C — Approach 3: Own the Design, AI as Instructor
```
I need to split $2400 rent among 3 roommates. HERE IS MY METHOD:
split the common space cost equally, and split the rest proportional
to bedroom square footage (200/150/130).

Before computing anything: challenge my method. What's unfair about
it? What edge cases break it? What would each roommate argue? Then
show me one alternative method and the tradeoff — but the final
choice is mine.
```


> 🗒️ **The comparison that matters:**
>
> A gave me: … (could I defend these numbers to the roommate paying the most?)
>
> B decided ______ for me without asking. The decisions it made that were actually MINE to make: …
>
> C pushed back on my method by pointing out: … — and the edge case I hadn't considered: …
>
> 💡 **Lock it in:** after C, I could explain and defend the split to all three roommates. After A, I could …
>
> 👥 **Pair moment:** compare Conversation C with a neighbor. Did Claude challenge your identical method the same way? Different challenges to the same design = free design review.
