## Exercise 6 — What changes, and what gets returned?
**Section 6 · About 9 minutes**

Start with:

```python
items = ["mouse", "keyboard"]
```

For each operation below, predict **two things** before running it:

1. What will `items` contain afterward?
2. What value will be stored in `result`?

We are deliberately **not** using `.sort()` here; that appears in the homework.


````python
# Run each block separately after predicting.

items = ["mouse", "keyboard"]
result = items.append("monitor")

print("append -> items:", items)
print("append -> result:", result)

print()

items = ["mouse", "keyboard"]
result = items.extend(["monitor", "camera"])
print("extend -> items:", items)
print("extend -> result:", result)

print()

items = ["mouse", "keyboard"]
result = items.insert(1, "monitor")
print("insert -> items:", items)
print("insert -> result:", result)

print()

items = ["mouse", "keyboard"]
result = items.pop()
print("pop -> items:", items)
print("pop -> result:", result)
````

**Discuss:**

- Which methods changed the list?
- Which methods returned `None`?
- Which method both changed the list **and returned a useful value**?
