## 1.1 Opening files: `with open()` (5 min)

A file persists outside the program. The `with` statement closes the file automatically, even if an exception occurs.

| Mode | Meaning | If the file already exists |
|---|---|---|
| `"r"` | Read | Keeps contents; file must exist |
| `"w"` | Write | **Erases previous contents** |
| `"a"` | Append | Adds to the end |

Use `encoding="utf-8"` for text. The `file` object inside the `with` block is our connection to the file.

````python
with open("notes.txt", "w", encoding="utf-8") as file:
    file.write("AAPL,10\n")
    file.write("MSFT,5\n")
    file.write("GOOG,7\n")

with open("notes.txt", "a", encoding="utf-8") as file:
    file.write("NVDA,3\n")

print("Created and appended to notes.txt")
````

### Before reading files: newlines and `repr()` (4 min)

The `\n` in `"AAPL,10\n"` represents **one newline character**. When `print()` encounters it, printing moves to the next line.

**New function: `repr()`** makes hidden characters visible in a string's representation. It helps us inspect text: `print()` displays the line break; `print(repr(...))` shows `\n` instead. It is a debugging/inspection tool, not something we need to use in ordinary file processing.

Compare the two results:

````python
message = "AAPL,10\nMSFT,5"
print("Normal print:")
print(message)
print("Using repr:")
print(repr(message))
````

## 1.2 Four ways to read a text file (8 min)

Each example opens `notes.txt` again so reading starts at the beginning.

- `read()` gets **all remaining text** in one string.
- `readline()` gets **one line** as a string, including the ending `\n` if present.
- `readlines()` gets **a list of strings**, one for each line.
- `for line in file:` processes **one line at a time**.

**Predict:** Which operations return a string and which return a list? We will use `repr()` to make the newline characters visible.

````python
with open("notes.txt", "r", encoding="utf-8") as file:
    whole_file = file.read()
print("read():", repr(whole_file), type(whole_file))

with open("notes.txt", "r", encoding="utf-8") as file:
    first = file.readline()
    second = file.readline()
print("readline():", repr(first), repr(second))

with open("notes.txt", "r", encoding="utf-8") as file:
    all_lines = file.readlines()
print("readlines():", all_lines, type(all_lines))

with open("notes.txt", "r", encoding="utf-8") as file:
    for line in file:
        print("for:", repr(line))
````

### Clean and split a line (3 min)

A line read from a file may end in `\n`. `strip()` removes whitespace at both ends of a string (including `\n`). `split(",")` splits a string into a **list** of fields at each comma.

Compare the results before we use this technique to rebuild positions from CSV.

````python
line = "AAPL,10\n"
print("Original:", repr(line))
print("Split only:", line.split(","))
print("After strip:", repr(line.strip()))
print("Strip, then split:", line.strip().split(","))
````

### The file remembers its reading position (2 min)

Each read advances the current position. When `read()` reaches the end, a second `read()` returns the **empty string** `""`, since no characters remain. `repr()` lets us see that empty string. Opening the file again starts from the beginning.

**Predict** the second result before you run the next cell.

````python
with open("notes.txt", "r", encoding="utf-8") as file:
    print("First read:", repr(file.read()))
    print("Second read:", repr(file.read()))  # Nothing remains
````

## 1.3 Five ways to write text (8 min)

Each example creates the **same file contents** in `write_demo.txt`. Run them one at a time: opening a file with `"w"` **replaces** its previous contents. All examples use `with open()`.

Compare how the newlines are handled. `write()` and `writelines()` do **not** automatically add `\n`; `print(..., file=file)` does.

**Method 1 — Multiple `write()` calls:** Explicitly write each line.

````python
with open("write_demo.txt", "w", encoding="utf-8") as file:
    file.write("AAPL,10\n")
    file.write("MSFT,5\n")
    file.write("GOOG,7\n")
````

**Method 2 — One `write()` call:** Put all the text in a single string.

````python
with open("write_demo.txt", "w", encoding="utf-8") as file:
    file.write("AAPL,10\nMSFT,5\nGOOG,7\n")
````

**Method 3 — `writelines()`:** Write a list of strings. The newlines must already be in the strings.

````python
lines = ["AAPL,10\n", "MSFT,5\n", "GOOG,7\n"]
with open("write_demo.txt", "w", encoding="utf-8") as file:
    file.writelines(lines)
````

**Method 4 — `print(..., file=file)`:** Write to the file instead of the screen. `print()` adds a newline by default.

````python
with open("write_demo.txt", "w", encoding="utf-8") as file:
    print("AAPL,10", file=file)
    print("MSFT,5", file=file)
    print("GOOG,7", file=file)
````

**Method 5 — A loop with `write()`:** Useful when the data already exists in a list.

````python
positions_text = ["AAPL,10", "MSFT,5", "GOOG,7"]
with open("write_demo.txt", "w", encoding="utf-8") as file:
    for position in positions_text:
        file.write(position + "\n")
````

**Appending:** Use `"a"` to add data without erasing the existing file. This works with any of the writing methods above.

````python
with open("write_demo.txt", "a", encoding="utf-8") as file:
    print("NVDA,3", file=file)

with open("write_demo.txt", "r", encoding="utf-8") as file:
    print(file.read())
````

## 1.4 CSV: rebuild a Python object by hand (10 min)

We receive `positions.csv` with a header `ticker,shares,avg_price`. Our **goal** is:

```python
positions = [
    {"ticker": "AAPL", "shares": 40, "avg_price": 182.50},
    # ... nine more dictionaries
]
```

A CSV row is text, not a Python dictionary. We must **skip the header, clean each line, split into fields, convert types, construct a dictionary, and append to a list**. We intentionally do **not** use the `csv` library today: we want to understand the work involved. For real-world CSV (quoted commas, etc.), use Python's `csv` module.

````python
positions = []

with open("positions.csv", "r", encoding="utf-8") as file:
    header = file.readline()  # Skip the header record
    print("Header:", repr(header))

    for line in file:
        fields = line.strip().split(",")
        ticker = fields[0].strip().upper()
        shares = int(fields[1])       # CSV values are text
        avg_price = float(fields[2])  # Must convert to a number

        position = {
            "ticker": ticker,
            "shares": shares,
            "avg_price": avg_price,
        }
        positions.append(position)

print("Number of positions:", len(positions))
print("First position:", positions[0])
print("Types:", type(positions[0]["shares"]),
      type(positions[0]["avg_price"]))
````

**Trace one row:** Starting with the raw text `AAPL,40,182.5\n`, write down (a) the output from `strip()`, (b) the list from `split(",")`, (c) the types before conversion, and (d) the resulting dictionary.

**What makes this fragile?** Think about an extra column, a missing value, invalid numbers, and commas inside quoted fields. We will *not* solve every CSV edge case today.

## 1.5 JSON: save and reconstruct the same object (7 min)

CSV required us to define the structure and convert each field. JSON records the **nested structure and primitive value types**, so the Python JSON library can reconstruct a list of dictionaries for us.

- `json.dump(object, file)` writes JSON to a file.
- `json.load(file)` reads JSON from a file and reconstructs its lists, dictionaries, strings, and numbers.
- `indent=2` makes the file readable. JSON supports common data structures, **not every Python object**.

````python
import json

with open("positions.json", "w", encoding="utf-8") as file:
    json.dump(positions, file, indent=2)

with open("positions.json", "r", encoding="utf-8") as file:
    restored_positions = json.load(file)

print("Same data?", restored_positions == positions)
print("Restored first position:", restored_positions[0])
print("Restored types:", type(restored_positions),
      type(restored_positions[0]),
      type(restored_positions[0]["shares"]),
      type(restored_positions[0]["avg_price"]))
````

**Contrast:** To reconstruct `positions` from CSV, count the parsing/conversion/assembly steps. To reconstruct it from JSON, how many steps did *we* write? Why does JSON work better for our list of dictionaries? Does JSON eliminate the need to validate data supplied by someone else?

# Part 2 — ISTI: External Inputs, Problem State, Program State (25 minutes)

The ISTI model: **Inputs → State → Transitions → Invariants**. Today we focus on external inputs and two useful views of state:

- **External input:** Information supplied from outside the algorithm, e.g., a user or a file.
- **Problem state:** The values and evolving information that represent the *problem we are solving*.
- **Program state:** Information that tracks **where we are in the algorithm**, such as the current loop index or current record being processed.

These are *conceptual roles*, not three mutually exclusive kinds of Python objects. A value can originate as external input and then form part of the problem state.

## 2.1 Midterm problem: Fibonacci (7 min)

**Task:** Given `n`, build a list containing the first `n` Fibonacci numbers.

Before running, identify the **external input**, **problem state**, and **program state**. Stop after a few iterations to see how each changes.

````python
n = 10                  # External input: number requested
fib = [0, 1]            # Problem state: sequence built so far

for idx in range(2, n): # Program state: current position in algorithm
    fib.append(fib[idx - 1] + fib[idx - 2])
    print("idx =", idx, "fib =", fib)

print("Result:", fib)
````

**Our distinction:**

| Role | Fibonacci example |
|---|---|
| External input | `n` (requested length) |
| Problem state | `n` and the evolving `fib` list |
| Program state | `idx`: which element we are calculating now |

**Discuss:** If we stop after `idx == 5`, what do we know about the sequence? What does `idx` tell us that `fib` does not directly tell us? Why is `n` both an external input and part of the problem state?

## 2.2 Midterm problem: Trades (8 min)

Each incoming trade is a string. We process trades in order to determine shares held by ticker. A malformed share count must be skipped. Identify what changes about the *problem* and what tracks our *position in processing*.

````python
trades = [
    "BUY AAPL 10", "BUY msft 5", "SELL AAPL 4",
    "BUY nvda ten", " buy aapl 3 ", "SELL MSFT 3", "BUY GOOG 7",
]

holdings = {}                    # Problem state: shares held so far
for idx, raw_trade in enumerate(trades):  # Program state: current trade index
    parts = raw_trade.strip().upper().split()
    action, ticker, quantity_text = parts
    try:
        shares = int(quantity_text)
    except ValueError:
        print("Skipped:", " ".join(parts))
        continue

    if action == "BUY":
        holdings[ticker] = holdings.get(ticker, 0) + shares
    elif action == "SELL":
        holdings[ticker] = holdings.get(ticker, 0) - shares
    print("After trade index", idx, ":", holdings)

print("Final holdings:")
for ticker in sorted(holdings):
    print(ticker, holdings[ticker])
````

| Role | Trades example |
|---|---|
| External input | The supplied list of trade strings |
| Problem state | Trades to process and the evolving `holdings` dictionary |
| Program state | `idx`: which trade is being processed now |

**Discuss:** Would the holdings alone tell you which trade you are processing? If an invalid trade is skipped, how does problem state differ from program state?
