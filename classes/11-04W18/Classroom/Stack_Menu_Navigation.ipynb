# Stack Menu Navigation

Run the program. Type a number to go into a menu, and press Enter to go back. Watch the stack printed after each choice.

````python
nav_menus = {
    "main": {
        "options": [
            {"menu_item": "Settings", "next_menu": "settings"},
            {"menu_item": "Help",     "next_menu": "help"},
        ]
    },
    "settings": {
        "options": [
            {"menu_item": "Display", "next_menu": None},
            {"menu_item": "Sound",   "next_menu": None},
        ]
    },
    "help": {
        "options": [
            {"menu_item": "FAQs",            "next_menu": None},
            {"menu_item": "Contact Support", "next_menu": None},
        ]
    },
}

# section 0
start_menu = list(nav_menus)[0]
stack = []
stack.append(start_menu)

while stack:
    # section 1
    cur_menu = nav_menus[stack[-1]]
    for idx, option in enumerate(cur_menu['options'], 1):
        print(f"{idx}) {option['menu_item']}")
    print()

    # section 2
    txt = 'Enter your choice: '
    while True:
        choice = input(txt)
        if choice == '':
            break
        try:
            choice = int(choice)
            if 1 <= choice <= len(cur_menu['options']):
                break
            else:
                raise ValueError
        except ValueError:
            txt = 'Enter a valid choice: '

    # section 3
    if choice == '':
        stack.pop()
    else:
        option = cur_menu['options'][choice - 1]    
        if option['next_menu']:   # what happens if this is None? 
            stack.append(option['next_menu'])
    
    print(stack)
````

### 6 things about `nav_menus` that need to be true

 1) The first key in `nav_menus` is our start menu - design choice.
 2) Each key in `nav_menus` refers to a menu dictionary.
 3) Every menu is a dictionary with an `'options'` key.
 4) `menu['options']` is a list of option dictionaries.
 5) Every option dictionary has keys `'menu_item'` and `'next_menu'`.
 6) Every `'next_menu'` key refers to either `None` or a key in `nav_menus`. !! IMPORTANT
