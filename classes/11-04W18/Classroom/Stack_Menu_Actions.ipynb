# Stack Menu Actions

In notebook 1, options could only move between menus. Now options can also *do* something: run an action.

### A function in a dictionary

A function name without parentheses is the function itself, a value like any other. That means we can store it in a dictionary and call it later by adding `()`.

```python
def show_faqs():
    print("Q: How do I go back?  A: Press Enter.")

actions = {"show_faqs": show_faqs}

actions["show_faqs"]()
```

Why do this?

- **It's the same idea as `nav_menus`.** `nav_menus` turns a name into a menu; `actions` turns a name into behavior. The menu only needs to store the name.
- **Names can be saved; functions can't.** A string like `"show_faqs"` can be written to a JSON file. The function itself can't, so the menu stores the name and `actions` looks it up.

### The program with actions

Each option now has a third key, `'action'`, which is either `None` or a key in `actions`. Only section 3 of the loop changes.

```python
def show_faqs():
    print("Q: How do I go back?  A: Press Enter.")

def not_implemented():
    print("NOT IMPLEMENTED")

def confirm():
    answer = input("Are you sure? (y/n) ")
    if answer != 'y':
        return False

actions = {
    "show_faqs":       show_faqs,
    "not_implemented": not_implemented,
    "confirm":         confirm,
}

nav_menus = {
    "main": {
        "options": [
            {"menu_item": "Settings", "next_menu": "settings", "action": None},
            {"menu_item": "Help",     "next_menu": "help",     "action": None},
            {"menu_item": "Advanced", "next_menu": "advanced", "action": "confirm"},
        ]
    },
    "settings": {
        "options": [
            {"menu_item": "Display", "next_menu": None, "action": "not_implemented"},
            {"menu_item": "Sound",   "next_menu": None, "action": None},
        ]
    },
    "help": {
        "options": [
            {"menu_item": "FAQs",            "next_menu": None, "action": "show_faqs"},
            {"menu_item": "Contact Support", "next_menu": None, "action": "not_implemented"},
        ]
    },
    "advanced": {
        "options": [
            {"menu_item": "Reset", "next_menu": None, "action": "not_implemented"},
        ]
    },
}

# section 0
start_menu = list(nav_menus)[0]
stack = []
stack.append(start_menu)

while stack:
    # section 1
    cur_menu = nav_menus[stack[-1]]
    for idx, option in enumerate(cur_menu['options'], 1):
        print(f"{idx}) {option['menu_item']}")
    print()

    # section 2
    txt = 'Enter your choice: '
    while True:
        choice = input(txt)
        if choice == '':
            break
        try:
            choice = int(choice)
            if 1 <= choice <= len(cur_menu['options']):
                break
            else:
                raise ValueError
        except ValueError:
            txt = 'Enter a valid choice: '

    # section 3
    if choice == '':
        stack.pop()
    else:
        option = cur_menu['options'][choice - 1]
        # section 4
        result = None
        if option['action']:
            result = actions[option['action']]()
        if option['next_menu'] and result is not False:
            stack.append(option['next_menu'])

    print(stack)
```

### What each kind of option does

| `menu_item` | `action` | `next_menu` | Result |
| --- | --- | --- | --- |
| Settings | `None` | `"settings"` | navigates |
| Display | `"not_implemented"` | `None` | acts, stays |
| Advanced | `"confirm"` | `"advanced"` | acts, then navigates unless the action returns `False` |
| Sound | `None` | `None` | stays |

Try each one.

### 8 things that need to be true

 1) The first key in `nav_menus` is our start menu - design choice.
 2) Each key in `nav_menus` refers to a menu dictionary.
 3) Every menu is a dictionary with an `'options'` key.
 4) `menu['options']` is a list of option dictionaries.
 5) Every option dictionary has keys `'menu_item'`, `'next_menu'` and `'action'`.
 6) Every `'next_menu'` key refers to either `None` or a key in `nav_menus` !! IMPORTANT
 7) Every `'action'` key refers to either `None` or a key in `actions` !! IMPORTANT
 8) An action can return `False` to cancel its option's `next_menu`.
