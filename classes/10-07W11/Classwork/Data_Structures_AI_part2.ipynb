## Part 1 — Grouping: Accumulating Into Buckets
**Target time: 25 minutes**

Suppose we already have a flat list of transaction records.

Management asks:

> "Break this down by region."

The blank in **"broken down by ___"** often identifies the grouping key.

The grouping pattern builds a dictionary:

- keys → bucket names
- values → lists of matching records

````python
# DATA FOR TODAY
from pprint import pprint as pp

transactions = [
    {"id": 1001, "store": "Boston",    "region": "East",  "category": "Shoes",   "amount": 240},
    {"id": 1002, "store": "Cambridge", "region": "East",  "category": "Apparel", "amount": 180},
    {"id": 1003, "store": "Albany",    "region": "North", "category": "Shoes",   "amount": 310},
    {"id": 1004, "store": "Boston",    "region": "East",  "category": "Gear",    "amount": 125},
    {"id": 1005, "store": "Hartford",  "region": "South", "category": "Apparel", "amount": 220},
    {"id": 1006, "store": "Albany",    "region": "North", "category": "Gear",    "amount": 95},
    {"id": 1007, "store": "Hartford",  "region": "South", "category": "Shoes",   "amount": 275},
    {"id": 1008, "store": "Cambridge", "region": "East",  "category": "Gear",    "amount": 150},
]

for transaction in transactions:
    print(transaction)
````

````python
# DESIGN BEFORE RUNNING
#
# Management asks:
#     "Show me the transactions broken down by region."
#
# Answer before coding:
#
# 1. What should the result dictionary's keys be?
# 2. What should each value contain?
# 3. Should each bucket keep ONLY amounts or WHOLE records?
# 4. What later questions become impossible if we keep only totals?
````

````python
# EXAMPLE — grouping whole records

by_region = {}

for transaction in transactions:
    key = transaction["region"]

    if key not in by_region:
        by_region[key] = []

    by_region[key].append(transaction)

for region in by_region:
    print("\n", region)
    for transaction in by_region[region]:
        print(" ", transaction)

````

## Part 2 — Counting: The Bucket Is a Number
**Target time: 15 minutes**

Counting has the same overall pattern as grouping:

1. initialize a result dictionary
2. loop through the source data
3. determine a key
4. update that key's bucket

The bucket type changes:

- grouping → bucket is a **list**
- counting → bucket is a **number**

`dict.get(key, 0)` is useful for counts.

````python
category_counts = {}

for transaction in transactions:
    key = transaction["amount"]
    category_counts[key] = category_counts.get(key, 0) + 1

print(category_counts)
````

## Part 3 — Fast Lookup: A Dictionary as an Index
**Target time: 20 minutes**

A list of records is good for:

- iteration
- preserving order
- sorting and filtering

But repeated requests such as:

> "Find the transaction with ID 1005."

do not require rescanning the list every time.

We can build an **index** once:

```text
id → complete record
```

That trades a small setup cost for cheap repeated lookup.

````python
transactions_by_id = {}

for transaction in transactions:
    transactions_by_id[transaction["id"]] = transaction

print(transactions_by_id[1005])
print(transactions_by_id[1005]["amount"])
````

## Part 4 — Choosing Among Shapes
**Target time: 20 minutes**

Common signals:

- many records with common fields → **list of dicts**
- rows and columns → **list of lists**
- things separated into buckets → **dict of lists**
- counts by key → **dict of counts**
- repeated lookup by a unique key → **dict index**
- one key mapped to a complete record → **dict of dicts**

There is rarely one universally correct representation.

A stronger design habit is:

1. choose a structure
2. name a reasonable alternative
3. say what each makes easy or expensive
4. state the invariant(s)

## Part 6 — Directing AI With a Contract
**Target time: 30 minutes**

A vague request leaves many decisions to the model.

A useful AI contract specifies the decisions that affect correctness or behavior:

- the **shape** of the input data
- required **fields / keys**
- what the function **accepts**
- what it **returns**
- whether inputs may be **mutated**
- relevant **edge cases**
- whether anything should be **printed**
- important invariants

The prompt is not where design starts.

> **The prompt documents decisions already made.**

## End-of-Class Reflection

Recognize these signals:

- **"broken down by..."** → grouping
- **"how many of each..."** → counting
- **"find by key repeatedly..."** → index
- **"choice maps to action..."** → dispatch dictionary
- **"repeat until quit..."** → menu loop
- **multiple structures describe the same reality** → define invariants
- **"ask AI to build it..."** → specify representation and behavior first

The full workflow is:

> **problem → questions → representation → invariants → pattern → AI contract / code → decision audit**

Good structure makes the next operation easier to reason about.  
Good specifications make the remaining delegation deliberate.
