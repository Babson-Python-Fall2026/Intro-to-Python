## Exercise 2 — Change the Implementation, Keep the Interface

Now the building data is stored differently:

```python
buildings = [
    ["HORN", "Horn Library"],
    ["GERB", "Gerber Hall"],
    ["OLIN", "Olin Hall"]
```

### Your task

Rewrite the **body** of `get_building(code)` so it works with the list of lists.

Use a basic `for` loop. Do **not** change the way the function is called:

```python
print(get_building("OLIN"))
```

Expected result:

```text
Olin Hall
```

The important idea is that the **implementation changes**, but the **caller does not have to change**.

````python
buildings = [
    ["HORN", "Horn Library"],
    ["GERB", "Gerber Hall"],
    ["OLIN", "Olin Hall"]
]

def get_building(code):
    for building in buildings:
        if building[0] == code:
            return building[1]

# The caller uses the function the same way as before.
print(get_building("OLIN"))
````

> **Guiding question:** *What are the useful pieces of the problem?*

A decomposed piece does **not** have to become a function. But functions are often a useful way to implement the pieces.

> **Key distinction**
>
> - **Decomposition:** separating the problem into pieces.
> - **Abstraction:** hiding the details of a piece behind a simpler interface.

## Exercise 9 — Implement the Decomposition Without Functions

Now implement the **entire grade-report problem** using the decomposition from Exercise 8.

For each student:

1. create the student's full name
2. calculate the quiz average
3. calculate the lab average
4. calculate the weighted course grade
5. create and store the result

### Important

Do **not** define any functions yet.

The goal is to show that a problem can be **decomposed into useful tasks even when the code is written inline**.

As you write the code, use comments to show where each decomposed task appears.


````python
from pprint import pprint

grades = []

for student in students:

    # 1. create the student's full name
    name = student["fname"] + " " + student["lname"]

    # 2. calculate the quiz average
    quiz_total = 0
    for score in student["quiz"]:
        quiz_total += score
    quiz_average = quiz_total / len(student["quiz"])

    # 3. calculate the lab average
    lab_total = 0
    for score in student["labs"]:
        lab_total += score
    lab_average = lab_total / len(student["labs"])

    # 4. calculate the weighted course grade
    final_grade = round(.3 * quiz_average + .7 * lab_average, 2)

    # 5. create and store the result
    result = {
        "name": name,
        "grade": final_grade
    }

    grades.append(result)

pprint(grades, sort_dicts=False)
````

# Part 3 — Pattern Recognition

> **Guiding question:** *What similarities, regularities, or repeated structures can I recognize?*

Pattern recognition is broader than recognizing shapes or pictures. In programming, we look for patterns in **values, data structures, code, behavior, and change over time**.

> **Important:** Recognizing a pattern is not the same as having an algorithm. First identify what repeats or changes. Then describe the rule precisely enough that a loop or algorithm could reproduce it.


---
## Common Types of Pattern Recognition

Here are some common kinds of patterns we encounter in computational thinking and programming.

| Pattern type | What we look for | Programming connection |
|---|---|---|
| **Numeric sequences** | How values change from one value to the next | Updating a variable inside a loop |
| **Repetition** | The same action applied again and again | `for` and `while` loops |
| **Data-structure patterns** | Items that have the same shape or keys | Processing collections consistently |
| **Code patterns** | Similar blocks of code doing the same kind of work | Loops, functions, or other reusable logic |
| **Spatial / visual patterns** | Symmetry, position, growth, shrinking | Graphics, layouts, grids, nested loops |
| **Input → output patterns** | A predictable relationship between input and result | Formulas, rules, functions |
| **Cycles / alternation** | A pattern that repeats through states | State changes, schedules, alternating behavior |
| **Patterns in real data** | Trends, seasonality, clusters, unusual values | Analysis, forecasting, anomaly detection |

A useful set of questions is:

- **What stays the same?**
- **What changes?**
- **By how much does it change?**
- **Does the change repeat, alternate, grow, shrink, or reverse?**
- **Can I describe the next step from the current step?**
- **Does the same operation apply to many items?**

