### Lists

#### A simple list

```python
my_list = [1, 2, 3]
```

A list can also contain **other collections**, including lists and dictionaries:

```python
my_list = [
    1,
    [1, 2, 3],
    {
        "Laptop": 1200.00,
        "Monitor": 300.00,
        "Mouse": 40.00
    }
]
```

This list contains three items:

1. the integer `1`
2. the list `[1, 2, 3]`
3. a dictionary containing product names and prices

> **Key idea:** A list is not limited to simple values. Its items can themselves be other collections.

#### A list of lists

Now let's store each product as a two-item list:

```python
products = [
    ["Laptop", 1200.00],
    ["Monitor", 300.00],
    ["Mouse", 40.00]
]
```

`products` is the **outer list**. Each item inside it is another list:

```python
["Laptop", 1200.00]
["Monitor", 300.00]
["Mouse", 40.00]
```

#### Accessing nested items

```python
products[0]       # ["Laptop", 1200.00]
products[0][0]    # "Laptop"
products[0][1]    # 1200.00
```

### For Loops

Using a basic for loop we can access the data with:

```python
for inner_list in products:
    print(inner_list, inner_list[0], inner_list[1])
```

On each pass through the loop:

```text
inner_list[0]  → name
inner_list[1]  → price
```

The basic pattern is:

```python
for item in collection:
    # do something with item
```

The next cells review for loops with lists and dictionaries.

````python

# for loop example for a list:

my_list = [1, 'banana', [1,2,3]]

for item in my_list:
    print(item)
print()

for idx, item in enumerate(my_list):
    print(idx, item)
print()

for idx, item in enumerate(my_list):
    print(idx, item)
    if idx == 2:        
        for idx, inner_item in enumerate(item):
            print(idx, inner_item)
print()


````

````python
# for loop example for dictionary

student = {'fname': 'Stacy',
           'lname': 'Smith',
           'dob':20070526}

for key in student:
    print(key)
print()

for k in student.keys():
    print(k)
print()

for v in student.values():
    print(v)
print()

for k, v in student.items():
    print(k, v)
````

### What Is a Function?

A **function** is a named block of code that performs a task.

You've already used functions that Python provides:

```python
print()
len()
sum()
```

Python also lets us write our own.

#### Basic structure

```python
def function_name(parameter):
    # instructions
    return result
```

| Part | Meaning |
|---|---|
| `def` | Tells Python we are **defining** a function |
| `function_name` | The name we choose for the function |
| `parameter` | A variable that receives a value when the function is called |
| indented body | The instructions the function performs |
| `return` | Sends a result back to the caller |

> **Key idea:** Defining a function does **not** run it. It teaches Python what to do when the function is called.

The function's **signature** is its name and parameters. For example:

```text
get_price(product)
```
