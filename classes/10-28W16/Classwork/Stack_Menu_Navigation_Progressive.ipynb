## Step 1 — A stack is just a Python list

A stack follows **last in, first out (LIFO)**. The last item placed on the stack is the first one removed. For menu navigation, the **last item is the current menu**.

| Statement | Meaning |
|---|---|
| `stack[-1]` | Look at the last item without removing it |
| `stack.append(item)` | Add an item to the top |
| `stack.pop()` | Remove the item at the top |

**Run each of the next eight code cells separately, in order.** There is only **one Python statement per cell**.


````python
stack = ["Main"]
````

````python
print(stack[-1])             # Main (look at the top)
````

````python
stack.append("Clients")
````

````python
print(stack)                 # ['Main', 'Clients']
````

````python
stack.pop()
````

````python
print(stack)                 # ['Main']
````

````python
stack.pop()
````

````python
print(stack)                 # []
````

**What happened?** We did not create a menu yet. We only saw how a list can remember a path. `stack[-1]` **looks** at the current item, `append()` moves **forward**, and `pop()` moves **back**.

A Jupyter cell containing `stack.pop()` might display the removed string because a notebook displays the value of its final expression. It is still removing that string from the list.

## Step 2 — Two menus, using the structure we will keep

We want to move between two menus. The picture shows the **current stack** at each location. Choosing **Clients** moves forward; pressing **Enter** moves back.

```text
                 MAIN MENU
               stack = ["main"]

                 1. Clients
                      |
                   choose 1
                      v
                CLIENTS MENU
         stack = ["main", "clients"]

               1. Create Client
                      |
                  press Enter
                      v
                   MAIN MENU
               stack = ["main"]
```

**First, describe the menus as data.** Each menu has a title and a list of options. Each option has a label (`menu_item`) and the name of a submenu (`next_menu`). `None` means the choice does not open another menu.

We are using **the same menu structure we will use in Step 6**, rather than introducing a temporary system with menu-specific `if` statements.


````python
menus = {
    "main": {
        "title": "MAIN MENU",
        "options": [
            {"menu_item": "Clients", "next_menu": "clients"}
        ]
    },
    "clients": {
        "title": "CLIENTS MENU",
        "options": [
            {"menu_item": "Create Client", "next_menu": None}
        ]
    }
}
````

### Find the menu at the top of the stack

We will use **temporary variables** to make each lookup easy to see. Follow the value moving from left to right:

`stack[-1]` → `menu_name` → `menus[menu_name]` → current menu definition.


````python
stack = ["main"]
menu_name = stack[-1]
print(menu_name)             # main
````

````python
menu = menus[menu_name]
print(menu["title"])       # MAIN MENU
````

### Follow the first option

For this two-menu example, Main has only one option, at **list index 0**. We look up its destination instead of writing a rule like `if menu == "main": ...`.


````python
option = menu["options"][0]
print(option)                # Both the label and the destination
````

````python
next_menu = option["next_menu"]
print(next_menu)             # clients
````

````python
stack.append(next_menu)      # Enter the Clients menu
print(stack)                 # ['main', 'clients']
````

````python
stack.pop()                  # Return to Main
print(stack)                 # ['main']
````

**Notice the separation:** `menus` describes **what the choices are**; `stack` remembers **where we are**. Neither structure has to replace the other.

**Think:** Why is the top of the stack `"clients"` instead of the words `"Create Client"`?


## Step 3 — Add Reports by changing the data, not the navigation rules

Now Main has **two choices**. Either choice opens a different submenu:

```text
                           MAIN MENU
                        stack = ["main"]

                          1. Clients
                          2. Reports
                            /    \\
                     choose 1    choose 2
                         /          \\
                        v            v
                 CLIENTS MENU     REPORTS MENU
                stack = ["main",  stack = ["main",
                         "clients"]         "reports"]

                1. Create Client   1. Holdings
                         \\           /
                         press Enter
                              |
                              v
                           MAIN MENU
                        stack = ["main"]
```

**Important:** The stack follows **one path at a time**. `['main', 'reports']` is one path; `['main', 'clients']` is another. The two submenus are never automatically added together.

We keep exactly the same field names. There is simply one more option under Main, and one new menu definition.


````python
menus = {
    "main": {
        "title": "MAIN MENU",
        "options": [
            {"menu_item": "Clients", "next_menu": "clients"},
            {"menu_item": "Reports", "next_menu": "reports"}
        ]
    },
    "clients": {
        "title": "CLIENTS MENU",
        "options": [
            {"menu_item": "Create Client", "next_menu": None}
        ]
    },
    "reports": {
        "title": "REPORTS MENU",
        "options": [
            {"menu_item": "Holdings", "next_menu": None}
        ]
    }
}
````

### Try the new route manually

Main's **Reports** choice is now at **list index 1** (because Python indexes lists from zero). We retrieve the destination from the data, then use the same `append()` and `pop()` operations as before.


````python
stack = ["main"]
menu = menus[stack[-1]]
option = menu["options"][1]      # Reports (second choice)
print(option)
````

````python
stack.append(option["next_menu"])
print(stack)                      # ['main', 'reports']
````

````python
stack.pop()
print(stack)                      # ['main']
````

We added a third menu **without introducing a new navigation rule**. The same stack operations worked for Clients and Reports.


## Step 4 — Display any menu using the same loop

So far we selected options manually. The next problem is **displaying the choices**. The menu definitions already contain the titles and labels. We only need to read them.

Start by displaying Main:


````python
stack = ["main"]
menu = menus[stack[-1]]

print(menu["title"])
for number, option in enumerate(menu["options"], 1):
    print(f"{number}. {option['menu_item']}")
````

Now pretend the user entered Reports. We use **identical display statements**; only the stack changes.


````python
stack.append("reports")
menu = menus[stack[-1]]

print(menu["title"])
for number, option in enumerate(menu["options"], 1):
    print(f"{number}. {option['menu_item']}")
````

**Look closely:** There is no `if menu == "reports"` or `if menu == "clients"`. The data determines what gets printed.

`enumerate(..., 1)` numbers the choices starting at **1** for people. Python still stores the first choice at list index **0**. We account for that in Step 5.


## Step 5 — Process one user choice

We can now **display** a menu. The next addition is to read **one** choice and update the stack. This is a **single navigation step**, not yet a program that repeats automatically.

The few `if` statements below handle **different kinds of input** — Enter, a valid number, or an invalid choice. They do **not** check the name of a particular menu. These same conditions will remain in the final program.

Run the cell once, then inspect its printed stack. Run it again to experiment with a different choice.


````python
stack = ["main"]
menu = menus[stack[-1]]

print(f"\n{menu['title']}")
for number, option in enumerate(menu["options"], 1):
    print(f"{number}. {option['menu_item']}")

choice = input("Choice (Enter to go back): ")

if choice == "":
    stack.pop()                         # Go back
elif choice.isdigit() and 1 <= int(choice) <= len(menu["options"]):
    option = menu["options"][int(choice) - 1]
    next_menu = option["next_menu"]

    if next_menu is None:
        print("Selected:", option["menu_item"])
    else:
        stack.append(next_menu)        # Go forward
else:
    print("Invalid choice")

print("Stack after one choice:", stack)
````

**What has changed?** Instead of hard-coding what choice **1** means, we find the option in `menu["options"]` and read its `next_menu` field.

**Why subtract 1?** The displayed choice **1** corresponds to list index **0**.

**Why `None`?** `Create Client` and `Holdings` are not submenus. We simply print the selection for now. Executing real functions is a later lesson.

## Step 6 — Make it a complete stack-driven menu system

We now have every piece:

- A stack to remember our route (Step 1).
- Menu definitions that contain the available choices and destinations (Steps 2–3).
- Code that displays **any** menu (Step 4).
- Code that processes **one** selection (Step 5).

**The final change is to repeat Steps 4 and 5 while the stack is not empty.** The `menus` dictionary from Step 3 is reused without changes.

Run the program. Try Main → Reports → Enter → Clients → Enter → Enter. Watch the stack after each change.


````python
stack = ["main"]

while stack:
    menu = menus[stack[-1]]

    print(f"\n{menu['title']}")
    print("Stack:", stack)

    for number, option in enumerate(menu["options"], 1):
        print(f"{number}. {option['menu_item']}")

    choice = input("Choice (Enter to go back): ")

    if choice == "":
        stack.pop()
    elif choice.isdigit() and 1 <= int(choice) <= len(menu["options"]):
        option = menu["options"][int(choice) - 1]
        next_menu = option["next_menu"]

        if next_menu is None:
            print("Selected:", option["menu_item"])
        else:
            stack.append(next_menu)
    else:
        print("Invalid choice")

print("Program ended")
````

### Compare Step 5 with Step 6

**No menu-specific conditions were added.** We only put the same display-and-select logic inside `while stack:`. This is why adding Reports in Step 3 did not require rewriting the navigation system.

| Expression | Question it answers |
|---|---|
| `stack[-1]` | Where am I now? |
| `menus[stack[-1]]` | What menu should I display? |
| `menu["options"]` | Which choices are available? |
| `option["next_menu"]` | Where should this choice take me? |
| `stack.append(next_menu)` | How do I move forward? |
| `stack.pop()` | How do I return? |

**Two kinds of `if` statements are very different:**

- `if menu == "main": ...` contains a **special rule for a menu**. It would have to grow whenever we add menus. We avoided it.
- `if choice == "": ...` contains a **general rule for navigation**. It applies to *every* menu, so it belongs in the final system.

The stack and dictionary now have distinct jobs: **the stack tracks the navigation path**; **the dictionary describes what each menu contains and where its options lead**.
