## Exercise 9 — Function arguments and local names
**Sections 10–11 · About 16 minutes**

When an argument is passed to a function, the parameter is a **local variable** that refers to the same object.

Python does **not** automatically make a private copy.

### Part A — mutation


````python
def add_device(items):
    print("inside before:", items, "id:", id(items))
    items.append("camera")
    print("inside after: ", items, "id:", id(items))
    
devices = ["mouse", "keyboard"]


add_device(devices)

print("outside     : ", devices, "id:", id(devices))
````

### Part B — rebinding the local parameter


````python
def replace_devices(items):
    print("inside before:", items, "id:", id(items))
    items = ["camera"]
    print("inside after: ", items, "id:", id(items))

devices = ["mouse", "keyboard"]


replace_devices(devices)

# (devices, id(devices))
````

In A and B above, explain why one function affects the caller’s list and the other does not.

In your explanation, use each of these terms correctly:

- **object**
- **reference**
- **local name**
- **mutation**
- **rebinding**

Then answer:

```python
new_items = items
```

inside the function — does that create a new list?


````python
def investigate(items):
    new_items = items
    print("items is new_items:", items is new_items)

devices = ["mouse", "keyboard"]
investigate(devices)
````

### Part B — shallow-copy surprise

Predict:

```python
a = [[1, 2], [3, 4]]
b = a.copy()
b[0].append(99)
```

Questions:

1. Is `a is b` true?
2. Is `a[0] is b[0]` true?
3. What will happen to `a`?


````python
a = [[1, 2], [3, 4]]
b = a.copy()

print("same outer list?")
print("same first inner list?")

b[0].append(99)
````

### Part C — deep copy


````python
import copy

a = [[1, 2], [3, 4]]
c = copy.deepcopy(a)

print("same outer list?")
print("same first inner list?")

c[0].append(99)

print("a:")
print("c:")
````

Explain:

> A shallow copy creates a new outer collection, but nested mutable objects may still be shared.
