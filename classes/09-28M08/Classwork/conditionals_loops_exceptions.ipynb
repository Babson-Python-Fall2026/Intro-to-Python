## 1. Quick Review: Shallow Copy vs. Deep Copy

Assignment creates another reference to the same object. A **shallow copy** creates a new outer container but still shares nested mutable objects. A **deep copy** recursively copies nested objects so the copied structure can change independently.


````python
# Predict the output before running.
# Why does changing a[0] also affect b[0]?

a = [[1, 2], [3, 4]]
b = a.copy()

a[0].append(99)

print("a:", a)
print("b:", b)
print("same outer object?", a is b)
print("same inner objects?", a[0] is b[0] and a[1] is b[1])
````

### Importing libraries

Two common import patterns are:

```python
import library as alias
```

and

```python
from library import function as alias
```

With the first form, functions are accessed through the library name or alias. With the second form, the imported function can be called directly.


````python
# Two equivalent ways to import deepcopy.
# Run each example separately and notice how the function call changes.

import copy

a = [[1, 2], [3, 4]]
b = copy.deepcopy(a)
a[0].append(99)

print("using import copy")
print("a:", a)
print("b:", b)

print()

from copy import deepcopy

x = [[1, 2], [3, 4]]
y = deepcopy(x)
x[0].append(99)

print("using from copy import deepcopy")
print("x:", x)
print("y:", y)

# Using aliases:
print()

import copy as cp

a = [[1, 2], [3, 4]]
b = cp.deepcopy(a)
a[0].append(99)

print("using import copy as cp")
print("a:", a)
print("b:", b)

print()

from copy import deepcopy as dcp

x = [[1, 2], [3, 4]]
y = dcp(x)
x[0].append(99)

print("using from copy import deepcopy as dcp")
print("x:", x)
print("y:", y)

# Questions:
# When would a shallow copy be enough?
# When would deepcopy() be useful?
````

### De Morgan's Laws

Negating a compound Boolean expression changes both the operator and each condition:

```python
(not (A and B)) == ((not A) or (not B))
(not (A or B))  == ((not A) and (not B))
```

Truth tables let us verify that two Boolean expressions are equivalent for every possible combination of inputs.


````python
'''
Before running the code, complete these two tables.

TABLE 1
A      B         (A and B)    not(A and B) | (not A)   (not B)   not(A) or not(B)
False  False         ?             ?             ?         ?              ?
False  True          ?             ?             ?         ?              ?
True   False         ?             ?             ?         ?              ?
True   True          ?             ?             ?         ?              ?

TABLE 2
A      B         (A or B)     not(A or B)  | (not A)   (not B)   not(A) and not(B)
False  False         ?             ?             ?         ?               ?
False  True          ?             ?             ?         ?               ?
True   False         ?             ?             ?         ?               ?
True   True          ?             ?             ?         ?               ?

Which two columns in each table always match?
'''

values = [False, True]

print("TABLE 1")
print("A", "B", "(A and B)", "not(A and B)", "(not A)", "(not B)", "not(A) or not(B)")

for A in values:
    for B in values:
        print(
            A,
            B,
            A and B,
            not (A and B),
            not A,
            not B,
            (not A) or (not B)
        )

print()

print("TABLE 2")
print("A", "B", "(A or B)", "not(A or B)", "(not A)", "(not B)", "not(A) and not(B)")

for A in values:
    for B in values:
        print(
            A,
            B,
            A or B,
            not (A or B),
            not A,
            not B,
            (not A) and (not B)
        )
````

## 4. `for` Loops

A `for` loop processes each item in an iterable. Lists, strings, tuples, dictionaries, and `range` objects can all be iterated over.

Three useful loop-control statements are:

- `pass` — do nothing; execution continues normally.
- `continue` — skip the rest of the current iteration.
- `break` — leave the loop immediately.


````python
'''
Trace this loop by hand.

For each iteration, record:
number | total before | total after






What is printed at the end?
'''

total = 0

for number in [2, 4, 6]:
    total += number


print(total)
````

````python
'''
PASS does nothing.

Predict the output.
Why might pass be useful while developing a program?
'''

for number in range(1, 6):
    if number == 3:
        pass
    print(number)
````

````python
'''
CONTINUE skips the rest of the current iteration.

Predict the output.
Why is 3 not printed?
'''

for number in range(1, 6):
    if number == 3:
        continue
    print(number)
````

````python
'''
BREAK exits the loop immediately.

Predict the output.
Which values never get processed?
'''

for number in range(1, 6):
    if number == 3:
        break
    print(number)
````

## 6. `while` Loops

A `while` loop repeats as long as its condition remains true.

Unlike a `for` loop, a `while` loop does not automatically advance through a sequence.  
The program must eventually change something that can make the condition false, or the loop may run forever.


````python
'''
Trace the loop.

For each iteration record:
count before | count after

How many times does the body execute?
'''

count = 1

while count <= 5:
    print(count)
    count += 1
````

## 7. Exceptions

Some operations can fail even when the program's logic is otherwise reasonable. A `try` statement lets us attempt an operation and handle specific exceptions.

The complete structure is:

```python
try:
    # code that might fail
except SomeError:
    # runs if that error occurs
else:
    # runs only if no exception occurred
finally:
    # runs whether or not an exception occurred
```

### Common Python exceptions

| Exception | What it means | Simple example |
|---|---|---|
| `ValueError` | The value is not acceptable for the operation. | `int("abc")` |
| `TypeError` | An operation is used with the wrong type of object. | `"5" + 3` |
| `IndexError` | A sequence index does not exist. | `[10, 20][5]` |
| `KeyError` | A dictionary key does not exist. | `{"a": 1}["b"]` |
| `NameError` | A variable or name has not been defined. | `print(score)` |
| `ZeroDivisionError` | Division or modulo by zero was attempted. | `10 / 0` |
| `FileNotFoundError` | Python cannot find the requested file. | `open("missing.txt")` |


````python
'''
Predict what runs when the user enters:

1. 25
2. hello

Pay attention to which blocks execute:
try
except
else
finally
'''

try:
    age = int(input("Enter your age: "))
except ValueError:
    print("except: please enter a whole number")
else:
    print("else: conversion succeeded")
    print("age:", age)
finally:
    print("finally: this always runs")
````

### Raising an exception yourself

Python raises many exceptions automatically, but your own program can also use `raise` when a value violates a rule.

For example:

```python
raise ValueError("message")
```

The minimal pattern for defining a custom exception is:

```python
class MyError(Exception):
    pass
```

Then you can raise it with:

```python
raise MyError("some message")
```


````python
'''
This program requires an integer from 1 through 10.

Predict what happens for:
5
0
11
hello

Notice that ValueError can come from:
1. int() if conversion fails
2. our own raise statement if the number is outside the allowed range
'''

try:
    number = int(input("Enter an integer from 1 to 10: "))

    if number < 1 or number > 10:
        raise ValueError("number must be between 1 and 10")

except ValueError as error:
    print("Invalid input:", error)
else:
    print("Accepted:", number)
finally:
    print("Input check complete.")
````
