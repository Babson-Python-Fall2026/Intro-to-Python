### Exercise D3 — Diagnose and repair malformed data
The setup below deliberately changes the key `"transactions"` into the incorrect key `"transaction"` for client `101`. Inspect the supplied validator's error message. Then repair **`repaired_clients`**, keeping all client records and transaction data; do not modify `client_records` or the validator.

This is the same general type of error that can occur when a student types `"menu_items"` instead of `"menu_item"`.

```python
# SUPPLIED validation function — read it, but do not rewrite it.
def validate_client_records(records):
    valid = True
    if not isinstance(records, dict):
        print('Client records must be a dictionary')
        return False
    for client_id, record in records.items():
        if not isinstance(record, dict):
            print(f'Client {client_id}: not a dictionary')
            valid = False
            continue
        if not isinstance(record.get('name'), str):
            print(f'Client {client_id}: missing/invalid name')
            valid = False
        contact = record.get('contact')
        if not isinstance(contact, dict) or not {'email', 'phone'} <= set(contact):
            print(f'Client {client_id}: missing/invalid contact')
            valid = False
        if not isinstance(record.get('transactions'), list):
            print(f'Client {client_id}: missing/invalid transactions list')
            valid = False
        if 'advisor' not in record:
            print(f'Client {client_id}: missing advisor field')
            valid = False
    return valid

malformed_clients = deepcopy(client_records)
malformed_clients[101]['transaction'] = malformed_clients[101].pop('transactions')
print('Before repair (expected to fail):')

validate_client_records(malformed_clients)
```

```python
repaired_clients = deepcopy(malformed_clients)  # Keep this line.
# YOUR CODE — correct the key without losing the transactions.

repaired_clients[101]['transactions'] = repaired_clients[101].pop('transaction')

print('After your repair:')
print('Structure valid:', validate_client_records(repaired_clients))
```

The next setup cell makes a separate copy of the **original starting menus** (not your E4/E5 modifications) and **deliberately inserts two errors**: one invalid menu destination and one incorrect option-field name. Read the validator messages, explain the two problems in comments, and fix the **`repaired_menus`** copy.

**Do not delete the affected menu or option to hide a problem.** Keep the original menu choices and make both structures and references valid.

```python
# SUPPLIED validator; same basic contract as Notebook 1. Do not rewrite it.
def validate_menus(menus):
    valid = True
    if not isinstance(menus, dict):
        print('Menu data must be a dictionary.')
        return False
    if 'main' not in menus:
        print('Missing starting menu: "main"')
        valid = False
    for menu_name, menu in menus.items():
        if not isinstance(menu, dict):
            print(f'{menu_name}: menu definition must be a dictionary')
            valid = False
            continue
        if not isinstance(menu.get('title'), str) or not menu.get('title').strip():
            print(f'{menu_name}: missing or invalid "title"')
            valid = False
        options = menu.get('options')
        if not isinstance(options, list):
            print(f'{menu_name}: missing or invalid "options" list')
            valid = False
            continue
        for number, option in enumerate(options, 1):
            location = f'{menu_name}, option {number}'
            if not isinstance(option, dict):
                print(f'{location}: option must be a dictionary')
                valid = False
                continue
            label = option.get('menu_item')
            if not isinstance(label, str) or not label.strip():
                print(f'{location}: missing or invalid "menu_item"')
                valid = False
            if 'next_menu' not in option:
                print(f'{location}: missing "next_menu"')
                valid = False
                continue
            next_menu = option['next_menu']
            if next_menu is not None:
                if not isinstance(next_menu, str) or not next_menu.strip():
                    print(f'{location}: next_menu must be a menu name or None')
                    valid = False
                elif next_menu not in menus:
                    print(f'{location}: undefined next_menu: {next_menu}')
                    valid = False
    if valid:
        print('Menu structures and submenu references are valid.')
    return valid

# Use the untouched Section E starting menus, independent of E4 and E5.
broken_menus = deepcopy(original_menus)
broken_menus['client_workspace']['options'][0]['next_menu'] = 'transactons'
broken_menus['main']['options'][0]['menu_items'] = broken_menus['main']['options'][0].pop('menu_item')
print('Broken copy (expected to fail):')
validate_menus(broken_menus)
```
