## 18–29 min · Purpose and derivability: two dimensions

We classify information by **two different questions**:

1. **Purpose:** What is the information used for? **Problem state** describes the modeled system; **execution state** tracks an algorithm's progress.
2. **Derivability:** Can it be calculated from the other information we retain? **Independent state** cannot; **derived state** can.

These are **dimensions**, not four rigid kinds of variables. What counts as derived depends on what other facts are retained, and derived information may still be stored for convenience.

**Example:** In ordinary Tic-Tac-Toe, board contents describe the problem. The current player and the winner can be derived from the board and rules; they are still facts about the current problem situation.

## 39–49 min · Program / execution state

A program has its own current situation while it works. **Execution state** tracks progress through a process; it need not represent a change in the underlying problem.

**Example — Searching**

```python
symbols = ["MSFT", "GOOG", "AAPL", "NVDA"]
target = "AAPL"
# A search is paused just before examining index 2.
```

We can keep the list and target unchanged while the search advances. The *search position* is execution state. It may be derivable from another progress record **if one exists**; it is not derivable from the unchanged list and target alone.

**Example — For loop:** The current item or index tells us where execution is. A completed-item counter might encode the same progress. Do not argue about the label of a variable; focus on its job.
