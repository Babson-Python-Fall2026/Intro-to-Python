**Definitions used throughout**
- **State:** information describing the current situation that is needed to determine what happens next.
- **Problem state:** information describing the system being modeled.
- **Execution state:** information describing an algorithm's progress while it works.
- **Independent / derived:** relative to other information retained; derived information can be calculated from that retained information.
- **Sufficient state:** together with the rules and a new input, it determines the permissible next behavior.

**Important:** Some information can play multiple roles. Your reasoning matters more than a label. When answering *derived or independent*, **say what other information is available**. A constant rule (such as garage capacity) is not necessarily changing state.
