````python
## 5. Trace a Loop — 5 points

numbers = [5, -2, 4, 0, 9]
total = 0
count = 0

for number in numbers:
    if number < 0:
        continue
    if number == 0:
        break
    total += number
    count += 1

print(total, count)

'''

A. (3 points) Complete the trace below in the cell. For the action, write `add`, `continue`, `break`, 
              or `not reached`. Record `total` and `count` after that value is handled; use `—` for the value 
              that is never reached.

| number | action | total after | count after |
|--------|--------|-------------|-------------|
|  5     |        |             |             |
| -2     |        |             |             |
|  4     |        |             |             |
|  0     |        |             |             |
|  9     |        |             |             |

B. (1 point) Give the final printed output.



C. (1 point) If only `continue` were replaced with `pass`, what would be printed? 
             Explain why the negative value would be treated differently.


 
'''
````

````python
# 6. Exception Handling — 5 points


for text in ["4", "bad", "0"]:
    try:
        number = int(text)
        result = 12 / number
    except ValueError:
        print("Invalid number")
    except ZeroDivisionError:
        print("Cannot divide by zero")
    else:
        print("Result", result)
    finally:
        print("Done")

'''

A. (1 point) Give the complete six-line output in order.



B. (2 points) For "bad" and "0", identify the exact statements that raise the exception and name the exception.



C. (2 points) Explain when `else` executes and when `finally` executes in this example.



'''
````
