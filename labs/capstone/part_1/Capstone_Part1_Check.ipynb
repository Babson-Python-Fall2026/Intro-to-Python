## Setup

Run this cell. It finds your PEMS folder and defines three helpers you'll use to look at your own files. You don't need to read the code.

````python
import ast
import json
from pathlib import Path

PEMS_DIR = Path("PEMS")
DATA = PEMS_DIR / "data"
SRC = PEMS_DIR / "source_code"

if not (DATA / "pems.json").exists() or not (SRC / "pems.py").exists():
    print("PEMS not found. This notebook must be in the folder that contains PEMS/ "
          "(labs/capstone/part_1).")
else:
    print("Found PEMS at", PEMS_DIR.resolve())


def show_pems(pm=None):
    """Print the clients and transactions in your pems.json (one PM, or all)."""
    with open(DATA / "pems.json", encoding="utf-8") as file:
        pems = json.load(file)
    for initials, record in pems.items():
        if pm is not None and initials != pm:
            continue
        clients = record["clients"]
        print(f"{initials}: {len(clients)} client(s)")
        for idx, client in enumerate(clients):
            print(f"  [{idx}] {client.get('name')}  ({len(client.get('transactions', []))} transaction(s))")
            for t in client.get("transactions", []):
                print(f"        #{t.get('trans_id')} day {t.get('trd_day')} {t.get('trans_type')} "
                      f"{t.get('ticker')} total={t.get('total')}")


def show_menus():
    """Print every option in your menus.json with its next_menu and action."""
    with open(DATA / "menus.json", encoding="utf-8") as file:
        menus = json.load(file)
    for name, menu in menus.items():
        print(f"{name}  ({menu['title']})")
        for number, option in enumerate(menu["options"], 1):
            print(f"  {number}) {option['menu_item']:16} next_menu={option.get('next_menu')!r:20} "
                  f"action={option.get('action', '<no action key>')!r}")


def show_function(name, filename="actions.py"):
    """Print one function from your source code exactly as it is in the file."""
    source = (SRC / filename).read_text(encoding="utf-8")
    for node in ast.parse(source).body:
        if isinstance(node, ast.FunctionDef) and node.name == name:
            print(ast.get_source_segment(source, node))
            return
    print(f"{name} not found in {filename}")
````
